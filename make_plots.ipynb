```python
import matplotlib.pyplot as plt
```

```python
# Data
answer_lengths = [23, 247, 455, 871, 934]
time_serial = [2.887, 35.699, 58.769, 170, 184]
time_distributed = [3.03, 16.168, 36.816, 73.15, 90.63]

# Plot time taken for serial vs distributed execution
plt.figure(figsize=(8, 5))
plt.plot(answer_lengths, time_serial, marker='o', color='gold', label="Serial Execution", linestyle='-')
plt.plot(answer_lengths, time_distributed, marker='s', color='orange', label="Distributed Execution (n=4)", linestyle='--')
plt.xlabel("Answer Length (tokens)")
plt.ylabel("Time Taken (seconds)")
plt.title("Execution Time: Serial vs Distributed Execution")
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Compute speedup
speedup = [s / d for s, d in zip(time_serial, time_distributed)]

# Plot speedup
plt.figure(figsize=(8, 5))
plt.plot(answer_lengths, speedup, marker='o', color='g', linestyle='-')
plt.xlabel("Answer Length (tokens)")
plt.ylabel("Speedup (Serial/Distributed)")
plt.title(f'Speedup of Distributed Execution (n=4). Average speedup = {round(sum(speedup)/len(speedup),2)}')
plt.grid(True)
plt.show()
```
